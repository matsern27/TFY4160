```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv("noice.csv")
nodes = ["node1", "node2", "node3"]

display(df)
```

```python
x_values = df["temp"]
for node in nodes:
    y_values = df[node]
    plt.plot(x_values, y_values,"--", label=f"{node}")
    plt.scatter(x_values, y_values)

plt.grid()
plt.legend()
plt.show()
```

```python
x_values = df["temp"]
for node in nodes:
    y_values = df[node]
    slopoe = np.gradient(x_values, y_values)
    plt.plot(x_values, slopoe)

plt.show()
```
